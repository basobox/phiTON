```python

#kategorisasi
#5-7 child ,11-17 adult, 18-99 teen, 100+ collector

newData = []

for i in df.agerange_min:
    #tambahkan kondisi untuk mengisi newData sesuai dengan kategori umur
    #1. age<10 -> child
    #2. 10 - 17 -> teen
    #3. age >= 18 -> adult

    if i < 10:
        newData.append("child") 
    elif 10 <= i <= 17:
        newData.append("teen")
    else:
        newData.append("adult")

df["age_category"] = newData

```

```python
df.aggregate({"US_retail_price": "mean", "agerange_min": "count"}) #menghitung rata-rata harga eceran (US_retail_price) dan jumlah potongan (pieces) dari DataFrame df dengan menggunakan metode `aggregate()`. Dengan menggunakan dictionary sebagai argumen, kita dapat menentukan fungsi agregasi yang akan diterapkan pada setiap kolom, sehingga memudahkan analisis atau visualisasi data yang berkaitan dengan harga dan jumlah potongan set Lego.
```

```python
#untuk rata rata harga 
df.aggregate({
    "US_retailPrice": 'mean',
    "agerange_min": 'count'
})
df.US_retailPrice.mean()
```
